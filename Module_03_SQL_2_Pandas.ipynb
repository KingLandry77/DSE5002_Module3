```python
# create the connection to the database

con=sqlite3.connect(infile)

#we need a database cursor as well
# some database systems do not require a separate cursor, just the connector

cur=con.cursor()
```

```python
# we want to see list of the available tables

#set up the query- this is an SQL language command- we will see more detail on these later

sql_tables_query="SELECT name FROM sqlite_master WHERE (type='table');"

cur.execute(sql_tables_query)
table_listing=cur.fetchall()

# this will return a list of all the tables in the database

# we can print out each table in the list using a list comprehension

[(table) for table in table_listing]
```

```python
# There is a nice way to do this, lets load the entire table into a data frame so we can easily work it

Tables_df=pd.read_sql_query("SELECT * FROM sqlite_master",con)
Tables_df
```

# What do we have here?

The sqlite_master table in a SQLite database has a listing of all the tables, functions, procedures, indices and fews stored
in the database.

Most databases have something similar, a table that keeps track of all the contents

*tables* are individual tables of data

an *index* is a hashed index or key value for a specific data tables.  An index speeds up the time it takes to access a 
specific row of a given table, but it takes up space in the database.   Typically there will be an index for the *key* or 
unique identifier for a give row.  We can create indices for other columns in the table (say a name or a phone number) 
that we know we will want to up frequently.

a *view* is a report or summary of the information contained in one or more tables.   A view is not stored like a table, but rather as the SQL commands needed to create the view.   This is really helpful for reports, summaries or questions that people often want to see.   The view acts like it is a table, but is stored as just commands, saving space and avoiding duplicating information.
